```python
import pandas as pd

# Load files
df = pd.read_csv("../../valuation/dataset_for_ml.csv")          
correct = pd.read_csv("../../vin-sphere test/short_vin/data_methaq(2.0).csv")     # File with correct makes

# Normalize chassis numbers
for d in (df, correct):
    d["chassisNumber"] = (
        d["chassisNumber"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

# Normalize makes in the correct file
correct["make"] = (
    correct["make"]
    .astype(str)
    .str.strip()
    .str.upper()
)

lookup = (
    correct[["chassisNumber", "make"]]
    .drop_duplicates("chassisNumber")
    .rename(columns={"make": "correct_make"})
)

df = df.merge(lookup, on="chassisNumber", how="left")

GROUPED_MAKES = {
    "TOYOTA",
    "CHERY",
    "GWM",
    "FAW",
    "GAC",
    "CHANGAN",
    "SAIC",
    "GEELY",
    "HARLEY-DAVIDSON",
    "PIAGGIO",
    "KTM",
}

mask = df["make"].isin(GROUPED_MAKES) & df["correct_make"].notna()

print("Rows to restore:", mask.sum())

df.loc[mask, "make"] = df.loc[mask, "correct_make"]

df.drop(columns="correct_make", inplace=True)

df.to_csv("make_restored.csv", index=False)
```

```python
import pandas as pd

dirty = pd.read_csv("make_restored.csv")
correct = pd.read_csv("../../vin-sphere test/short_vin/data_methaq(2.0).csv") 

# Normalize chassis numbers
for df in (dirty, correct):
    df["chassisNumber"] = (
        df["chassisNumber"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

# Normalize make
correct["make"] = (
    correct["make"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Create lookup ONLY for VOYAH rows
voyah_lookup = (
    correct[correct["make"] == "VOYAH"][["chassisNumber", "make"]]
    .drop_duplicates("chassisNumber")
    .rename(columns={"make": "correct_make"})
)

# Merge
dirty = dirty.merge(voyah_lookup, on="chassisNumber", how="left")

# Update only matched rows
mask = dirty["correct_make"].notna()
dirty.loc[mask, "make"] = "VOYAH"

print(f"Updated {mask.sum()} rows to VOYAH")

dirty.drop(columns="correct_make", inplace=True)
dirty.to_csv("dirty_updated.csv", index=False)
```
