```python
import pandas as pd
import glob

# 1. Get a list of all CSV file paths in your target folder
# Replace 'your_folder_path' with the actual path to your files
file_paths = glob.glob('../Valuation/data/drivearabia_car_depreciation_valuation/depreciated_2/*_dep.csv')

# 2. Use a list comprehension to read all CSVs into memory efficiently
df_list = [pd.read_csv(file) for file in file_paths]

# 3. Concatenate them all at once
# axis=0 stacks them vertically (row-wise)
combined_df = pd.concat(df_list, axis=0, ignore_index=True)

# 4. Optional: Save your new mega-dataset
combined_df.to_csv('../Valuation/data/drivearabia_car_depreciation_valuation/car_dep_2.csv', index=False)

print(f"Successfully merged {len(file_paths)} files!")
```
