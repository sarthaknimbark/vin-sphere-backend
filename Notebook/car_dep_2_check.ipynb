```python
import pandas as pd
```

```python
df = pd.read_csv("../Valuation/data/drivearabia_car_depreciation_valuation/car_dep_2.csv")
```

```python
df
```

```python
# Calculate the depreciated value using the formula
df['depreciated_value'] = df['price_avg_aed'] * (1 - df['annual_dep_rate']) ** df['car_age']

# Optional: Round the result to match the integer values shown in your dataframe
df['depreciated_value'] = df['depreciated_value'].round(0).astype(int)
```
