```python
df_master = pd.read_csv("../Data/master-comparison.csv")
df_lookup = pd.read_csv("../Data/lookup_clean_translated.csv")
```

```python
df_master_merge = df_master[["vin", "realMake", "realModel", "realTrim", "realValue", "realYear"]]
```

```python
df_master_merge
```

```python
df_master_merge = df_master_merge.rename(columns={
    "vin" : "chassisNumber",
    "realMake" : "make",
    "realModel" : "model",
    "realTrim" : "trim",
    "realValue" : "price",
    "realYear" : "year"
})
```

```python
# Set 'chassisNumber' as the index, update df1 with df2, and reset the index back
new_df = df_master_merge.set_index('chassisNumber').combine_first(df_lookup.set_index('chassisNumber')).reset_index()
```

```python
new_df = new_df.drop_duplicates()
```
